### Global Happiness Report

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
print("Happiness Analysis Started!")
```

```python
url = "https://raw.githubusercontent.com/plotly/datasets/master/2015-world-happiness-report.csv"

df = pd.read_csv(url)

print("Dataset Loaded Successfully!")
```

```python
print(df.head())
print("\nShape:", df.shape)
df.info()
```

```python
data = df.copy()

data = data.drop_duplicates()
data = data.fillna(data.median(numeric_only=True))

print("Missing Values:")
print(data.isnull().sum())
```

```python
top_countries = data.sort_values(
    "Happiness Score", ascending=False
).head(10)

print(top_countries[["Country", "Happiness Score"]])
```

```python
sns.barplot(
    data=top_countries,
    x="Happiness Score",
    y="Country"
)

plt.title("Top 10 Happiest Countries")
plt.show()
```

```python
print("Average Happiness:", data["Happiness Score"].mean())

print("Highest Happiness:",
      data.loc[data["Happiness Score"].idxmax(), "Country"])

print("Lowest Happiness:",
      data.loc[data["Happiness Score"].idxmin(), "Country"])

print("Happiness Analysis Completed!")
```

```python
data.to_csv("Titanic_Survival_Analysis.csv", index=False)

print("Titanic CSV File Saved Successfully!")
```
