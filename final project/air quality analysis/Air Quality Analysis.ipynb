## Air Quality Analysis

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
print("Air Quality Analysis Started!")
```

```python
url = "https://raw.githubusercontent.com/plotly/datasets/master/airquality.csv"

df = pd.read_csv(url)

print("Dataset Loaded!")
```

```python
print(df.head())
print("\nShape:", df.shape)
print("\nColumns:", df.columns)
df.info()
```

```python
data = df.copy()

data = data.drop_duplicates()

data = data.fillna(data.median(numeric_only=True))

print("Missing Values:")
print(data.isnull().sum())
```

```python
print("Average Ozone:", data["Ozone"].mean())
print("Average Temperature:", data["Temp"].mean())
print("Average Wind:", data["Wind"].mean())

print("\nStatistical Summary:")
print(data.describe())
```

```python
sns.lineplot(data=data, x="Day", y="Temp")

plt.title("Daily Temperature Analysis")
plt.xlabel("Day")
plt.ylabel("Temperature")

plt.show()
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 5))

sns.histplot(data=data, x="Ozone", kde=True, ax=ax[0])
ax[0].set_title("Ozone Distribution")

sns.scatterplot(data=data, x="Wind", y="Ozone", ax=ax[1])
ax[1].set_title("Wind vs Ozone")

plt.tight_layout()
plt.show()
```

```python
print("========== AIR QUALITY REPORT ==========")

print("Average Ozone:", round(data["Ozone"].mean(), 2))
print("Average Wind:", round(data["Wind"].mean(), 2))
print("Average Temperature:", round(data["Temp"].mean(), 2))

print("Air Quality Analysis Completed!")
```

```python
data.to_csv("Air_Quality_Analysis.csv", index=False)

print("Air Quality CSV File Saved Successfully!")
```
